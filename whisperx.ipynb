```python
from faster_whisper import WhisperModel

model = WhisperModel(
    "large-v3",
    device="cuda",
    compute_type="float16"
)

print("✅ Model Loaded")
```

```python
segments, info = model.transcribe(
    audio_file,
    beam_size=5,
    word_timestamps=True
)

print("Language:", info.language)
```

```python
import json

words = []
index = 1

for segment in segments:

    if segment.words is None:
        continue

    for w in segment.words:

        words.append({
            "index": index,
            "word": w.word.strip(),
            "start": round(w.start, 3),
            "end": round(w.end, 3),
            "probability": round(w.probability, 3)
        })

        index += 1

print("Total Words:", len(words))

with open("word_timestamps.json", "w", encoding="utf-8") as f:
    json.dump(
        words,
        f,
        ensure_ascii=False,
        indent=2
    )

print("✅ JSON Saved")
```

# **Image Rename**

```python
import os
import json
import shutil
import csv
import re

# ==========================================
# CONFIG
# ==========================================

JSON_FILE = r"storyboard.json"
IMAGE_FOLDER = r"Images"

SUPPORTED = (".png", ".jpg", ".jpeg", ".webp")

# ==========================================
# LOAD STORYBOARD
# ==========================================

with open(JSON_FILE, "r", encoding="utf-8") as f:
    data = json.load(f)

blocks = data["visual_blocks"]
expected = len(blocks)

# ==========================================
# LOAD IMAGES
# ==========================================

images = [
    f for f in os.listdir(IMAGE_FOLDER)
    if f.lower().endswith(SUPPORTED)
]

# ==========================================
# SORT USING BLOCK NUMBER
# ==========================================

def get_block_number(filename):
    match = re.search(r'Block_(\d+)', filename, re.IGNORECASE)

    if match:
        return int(match.group(1))

    return 999999


images.sort(key=get_block_number)

found = len(images)

print("=" * 60)
print("SMART IMAGE RENAMER")
print("=" * 60)

print(f"Expected Images : {expected}")
print(f"Found Images    : {found}")

# ==========================================
# VALIDATE
# ==========================================

if found != expected:

    print("\nERROR: Image count does not match storyboard.")

    if found < expected:
        print(f"Missing Images : {expected-found}")

    else:
        print(f"Extra Images : {found-expected}")

    print("\nPlease fix the mismatch before renaming.")
    exit()

choice = input("\nContinue Renaming? (y/n): ")

if choice.lower() != "y":
    exit()

# ==========================================
# BACKUP
# ==========================================

backup = os.path.join(IMAGE_FOLDER, "Backup_Original")

os.makedirs(backup, exist_ok=True)

for img in images:

    shutil.copy2(
        os.path.join(IMAGE_FOLDER, img),
        os.path.join(backup, img)
    )

print("\nBackup Created Successfully.")

# ==========================================
# TEMP RENAME
# ==========================================

temp_files = []

for i, img in enumerate(images):

    old_path = os.path.join(IMAGE_FOLDER, img)

    ext = os.path.splitext(img)[1]

    temp_name = f"temp_{i:05d}{ext}"

    temp_path = os.path.join(IMAGE_FOLDER, temp_name)

    os.rename(old_path, temp_path)

    temp_files.append(temp_name)

# ==========================================
# FINAL RENAME
# ==========================================

log = []

for block, temp in zip(blocks, temp_files):

    block_id = block["block_id"]

    ext = os.path.splitext(temp)[1]

    old_path = os.path.join(IMAGE_FOLDER, temp)

    new_name = f"{block_id:03d}{ext}"

    new_path = os.path.join(IMAGE_FOLDER, new_name)

    os.rename(old_path, new_path)

    log.append([
        temp,
        new_name,
        block["start_time"],
        block["end_time"],
        block["duration"]
    ])

# ==========================================
# CSV LOG
# ==========================================

csv_file = os.path.join(
    IMAGE_FOLDER,
    "rename_log.csv"
)

with open(csv_file, "w", newline="", encoding="utf-8") as f:

    writer = csv.writer(f)

    writer.writerow([
        "Old Name",
        "New Name",
        "Start Time",
        "End Time",
        "Duration"
    ])

    writer.writerows(log)

# ==========================================
# DONE
# ==========================================

print("\n" + "=" * 60)
print("SUCCESS")
print("=" * 60)

print(f"Images Renamed : {len(log)}")
print(f"Backup Folder  : {backup}")
print(f"CSV Log        : {csv_file}")

print("\nAll images have been renamed successfully.")
```

# **Rendering**

```python
import json
import subprocess
from pathlib import Path

import imageio.v2 as imageio
import imageio_ffmpeg
import numpy as np
from PIL import Image
from tqdm import tqdm

# ======================================================
# CONFIG
# ======================================================

WIDTH = 1920
HEIGHT = 1080
FPS = 30
BACKGROUND = (255, 255, 255)

SUPPORTED_EXTENSIONS = [
    ".png",
    ".jpg",
    ".jpeg",
    ".webp"
]

BASE = Path(__file__).resolve().parent

JSON_FILE = BASE / "storyboard.json"
AUDIO_FILE = BASE / "narration.mp3"
IMAGE_DIR = BASE / "images"
OUTPUT_DIR = BASE / "output"

OUTPUT_DIR.mkdir(exist_ok=True)

TEMP_VIDEO = OUTPUT_DIR / "video_silent.mp4"
FINAL_VIDEO = OUTPUT_DIR / "Final.mp4"

# ======================================================
# VALIDATION
# ======================================================

print("=" * 60)
print("Checking project files...")
print("=" * 60)

if not JSON_FILE.exists():
    raise FileNotFoundError(f"Missing:\n{JSON_FILE}")

if not AUDIO_FILE.exists():
    raise FileNotFoundError(f"Missing:\n{AUDIO_FILE}")

if not IMAGE_DIR.exists():
    raise FileNotFoundError(f"Missing folder:\n{IMAGE_DIR}")

# ======================================================
# LOAD STORYBOARD
# ======================================================

print("Loading storyboard...")

with open(JSON_FILE, "r", encoding="utf-8") as f:
    data = json.load(f)

blocks = data["visual_blocks"]

print(f"Found {len(blocks)} visual blocks")

# ======================================================
# IMAGE PREPARATION
# ======================================================

def prepare_image(path: Path):

    img = Image.open(path).convert("RGB")

    canvas = Image.new(
        "RGB",
        (WIDTH, HEIGHT),
        BACKGROUND
    )

    img.thumbnail(
        (WIDTH, HEIGHT),
        Image.Resampling.LANCZOS
    )

    x = (WIDTH - img.width) // 2
    y = (HEIGHT - img.height) // 2

    canvas.paste(img, (x, y))

    return np.array(canvas)

# ======================================================
# IMAGE FINDER
# ======================================================

def find_image(block):

    # --------------------------------------------------
    # image_file from JSON
    # --------------------------------------------------

    if block.get("image_file"):

        candidate = IMAGE_DIR / block["image_file"]

        if candidate.exists():
            return candidate

    # --------------------------------------------------
    # Exact filename
    # --------------------------------------------------

    prefix = f'{block["block_id"]:03d}'

    for ext in SUPPORTED_EXTENSIONS:

        candidate = IMAGE_DIR / f"{prefix}{ext}"

        if candidate.exists():
            return candidate

    # --------------------------------------------------
    # Descriptive filename
    # Example:
    # 001_intro.jpeg
    # 001_scene.webp
    # --------------------------------------------------

    for ext in SUPPORTED_EXTENSIONS:

        matches = list(
            IMAGE_DIR.glob(f"{prefix}_*{ext}")
        )

        if matches:
            return matches[0]

    # --------------------------------------------------
    # Any file beginning with block id
    # --------------------------------------------------

    matches = sorted(
        IMAGE_DIR.glob(f"{prefix}*")
    )

    for file in matches:

        if file.suffix.lower() in SUPPORTED_EXTENSIONS:
            return file

    return None

# ======================================================
# CREATE VIDEO
# ======================================================

print("\nCreating silent video...\n")

writer = imageio.get_writer(
    str(TEMP_VIDEO),
    fps=FPS,
    codec="libx264",
    quality=9
)

total_blocks = len(blocks)

missing = 0
used = 0

for i, block in enumerate(tqdm(blocks)):

    image_path = find_image(block)

    if image_path is None:

        print(
            f"Missing image for block "
            f'{block["block_id"]:03d}'
        )

        missing += 1
        continue

    try:

        frame = prepare_image(image_path)

    except Exception as e:

        print(
            f"Cannot open image:\n"
            f"{image_path}\n{e}"
        )

        missing += 1
        continue

    start_frame = round(
        float(block["start_time"]) * FPS
    )

    if i < total_blocks - 1:

        end_frame = round(
            float(blocks[i + 1]["start_time"]) * FPS
        )

    else:

        end_frame = round(
            float(block["end_time"]) * FPS
        )

    frame_count = max(
        1,
        end_frame - start_frame
    )

    for _ in range(frame_count):
        writer.append_data(frame)

    used += 1

writer.close()

print()

print("=" * 60)
print("SUMMARY")
print("=" * 60)
print(f"Blocks       : {total_blocks}")
print(f"Images Used  : {used}")
print(f"Missing      : {missing}")

# ======================================================
# VERIFY VIDEO
# ======================================================

if used == 0:

    raise RuntimeError(
        "No images were found.\n"
        "Please check the images folder."
    )

if not TEMP_VIDEO.exists():

    raise RuntimeError(
        "Silent video was not created."
    )

print("\nSilent video created successfully.")

# ======================================================
# MERGE AUDIO
# ======================================================

print("\nMerging audio...\n")

ffmpeg = imageio_ffmpeg.get_ffmpeg_exe()

cmd = [
    ffmpeg,
    "-y",
    "-i", str(TEMP_VIDEO),
    "-i", str(AUDIO_FILE),
    "-c:v", "copy",
    "-c:a", "aac",
    "-shortest",
    str(FINAL_VIDEO)
]

result = subprocess.run(
    cmd,
    capture_output=True,
    text=True
)

if result.returncode != 0:

    print(result.stderr)

    raise RuntimeError(
        "FFmpeg failed."
    )

if not FINAL_VIDEO.exists():

    raise RuntimeError(
        "Final video was not created."
    )

print()

print("=" * 60)
print("DONE!")
print("=" * 60)
print(f"Output Video : {FINAL_VIDEO}")
print("=" * 60)
```
